8. Write a Python program that accepts a string and prints the number of vowels in it. 


```python
text = input("enter string:")
vowels = 'aeiou'
count = sum(1 for char in text if char.lower() in vowels)
print('number of vowels:', count)
```

 16. **Question:** Write a program that checks if the input number is a prime 
number. 

```python
num = int(input("enter a number:"))
if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print("not prime number")
            break
        else:
            print("prime number")
```
