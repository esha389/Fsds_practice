```python
logical_test = np.array([True, True,True])
result = np.all(logical_test)
print(result)
```

```python
logical_test1 = np.array([True, False,True])
result = np.all(logical_test1) #if all elements are true then result is true
print(result)
```

```python
result1 = np.any(logical_test1)
print(result1) #if any element is true than its true
```

```python
set_a = np.array([1, 2, 3, 4])
set_b = np.array([3, 4, 5, 6])
intersection = np.intersect1d(set_a, set_b)
print("Intersection of a and b:", intersection)
```

```python
union = np.union1d(set_a,set_b)
print(union)
```
